This notebook demonstrates the data ingestion, cleaning, and exploratory analysis steps that feed the ML pipeline.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Reproducibility
np.random.seed(42)
sns.set(style='whitegrid')
```

## 1. Load Raw Datasets

```python
# Email sample
email_df = pd.read_csv('../data/raw/sample_emails.csv')
print(f'Email samples: {len(email_df)}')
display(email_df.head())
```

## 2. Load SMS Spam Collection

```python
sms_df = pd.read_csv('../data/raw/sample_sms.csv', names=['label', 'text'], sep='\t')
sms_df['label'] = sms_df['label'].map({'ham': 0, 'spam': 1})
print(f'SMS samples: {len(sms_df)} | Spam ratio: {sms_df["label"].mean():.2%}')
```

## 3. Text Length & Class Distribution

```python
sms_df['text_length'] = sms_df['text'].str.len()
plt.figure(figsize=(8, 4))
sns.histplot(data=sms_df, x='text_length', hue='label', bins=30, kde=True)
plt.title('Message Length Distribution by Class')
plt.xlabel('Characters')
plt.show()
```

## 4. Key Observations
- SMS spam messages tend to have shorter, more urgent language.
- Email spam contains URLs, currency symbols (`$`), and excessive capitalization.
- These observations motivate the handcrafted feature engineering in `src/features/`.
