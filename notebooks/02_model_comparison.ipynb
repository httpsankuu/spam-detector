# 02 - Model Comparison & Benchmark Analysis

Compares 5 ML algorithms (Naive Bayes, Logistic Regression, Linear SVM, Random Forest, XGBoost) on the held-out test split.

```python
import json
import pandas as pd
import joblib
import matplotlib.pyplot as plt
from pathlib import Path
```

## Load Benchmark Results

```python
with open('../reports/test_benchmark.json') as f:
    bench = json.load(f)
df = pd.DataFrame(bench['summary'])
df['Model'] = df['Model'].str.replace('_', ' ').str.title()
display(df[['Model', 'Accuracy', 'F1-Score', 'ROC-AUC']].sort_values('F1-Score', ascending=False))
```

## Model Load & Quick Inference

```python
models_dir = '../models'
model_names = ['logistic_regression', 'linear_svm', 'naive_bayes', 'random_forest', 'xgboost']
loaded = {}
for name in model_names:
    path = Path(models_dir) / f'{name}_pipeline.joblib'
    if path.exists():
        loaded[name] = joblib.load(path)
        print(f'Loaded: {name}')
print(f'Total pipelines loaded: {len(loaded)}')
```

## Why Logistic Regression Wins
- Linear model provides **transparent coefficients** (`x_i · w_i`) for explainability.
- Best F1-Score (~0.957) with lowest false negatives.
- Faster inference than tree ensembles, ideal for real-time Streamlit demo.
